# 06 启动中文 Gradio GUI

自动发现本项目实际训练产生的权重。优先读取 `models/` 部署目录，兼容 `weights/` 训练输出目录。没有真实权重时 GUI 仍可启动并显示中文提示，不回退到官方预训练模型。

```python
from pathlib import Path
import sys

def locate_project_root(start=Path.cwd()):
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate / 'src').is_dir() and (candidate / 'notebooks').is_dir():
            return candidate
    raise FileNotFoundError('找不到项目根目录')

PROJECT_ROOT = locate_project_root()
sys.path.insert(0, str(PROJECT_ROOT))
from src.app import build_demo
from src.cloud_training import available_trained_weights

MODEL_NAMES = ['YOLO11n', 'YOLO11s', 'YOLO26n', 'YOLO26s']
model_registry = available_trained_weights(
    MODEL_NAMES,
    PROJECT_ROOT / 'weights',
    models_root=PROJECT_ROOT / 'models',
    metrics_path=PROJECT_ROOT / 'reports' / 'model_comparison.csv',
)
demo = build_demo(model_registry)
demo.launch()
```
