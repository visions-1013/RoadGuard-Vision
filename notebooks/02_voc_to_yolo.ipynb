# 02 Pascal VOC 转 YOLO

按子集分别划分后合并，固定随机种子为 `42`。

```python
from pathlib import Path
import sys
sys.path.append('..')
from src.data_utils import convert_dataset, export_issues_csv, generate_data_yaml, merge_converted_splits

SOURCE_ROOT = Path('../datasets/RDD2022')
INTERMEDIATE_ROOT = Path('../data/rdd2022_intermediate')
YOLO_ROOT = Path('../data/rdd2022_yolo')
summary = convert_dataset(SOURCE_ROOT, INTERMEDIATE_ROOT)
export_issues_csv(summary['issues'], Path('../data/issues.csv'))
split_counts = merge_converted_splits(INTERMEDIATE_ROOT, YOLO_ROOT, seed=42)
generate_data_yaml(Path('../data/data.yaml'), YOLO_ROOT)
{'converted_samples': summary['converted_samples'], 'issues': len(summary['issues']), 'split_counts': split_counts}
```

```python
label_paths = list((YOLO_ROOT / 'labels').rglob('*.txt'))
for path in label_paths[:20]:
    for row in path.read_text(encoding='utf-8').splitlines():
        values = [float(value) for value in row.split()[1:]]
        assert all(0 <= value <= 1 for value in values), (path, row)
print(f'抽查完成：{min(20, len(label_paths))} 个标签文件')
```
