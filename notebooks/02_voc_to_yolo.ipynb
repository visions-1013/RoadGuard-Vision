# 02 Pascal VOC 转 YOLO

按子集分别划分后合并，固定随机种子为 `42`。默认重建生成目录，避免重复运行时保留旧标签。

```python
from pathlib import Path
import sys

def locate_project_root(start=Path.cwd()):
    for candidate in (start.resolve(), *start.resolve().parents):
        if (candidate / 'src').is_dir() and (candidate / 'notebooks').is_dir():
            return candidate
    raise FileNotFoundError('找不到项目根目录')

PROJECT_ROOT = locate_project_root()
sys.path.insert(0, str(PROJECT_ROOT))
from src.cloud_training import prepare_yolo_dataset

SOURCE_ROOT = PROJECT_ROOT / 'datasets' / 'RDD2022'
DATA_ROOT = PROJECT_ROOT / 'data'
summary = prepare_yolo_dataset(SOURCE_ROOT, DATA_ROOT, rebuild=True)
summary
```

```python
YOLO_ROOT = Path(summary['yolo_root'])
label_paths = list((YOLO_ROOT / 'labels').rglob('*.txt'))
for path in label_paths[:20]:
    for row in path.read_text(encoding='utf-8').splitlines():
        values = [float(value) for value in row.split()[1:]]
        assert all(0 <= value <= 1 for value in values), (path, row)
print(f'抽查完成：{min(20, len(label_paths))} 个标签文件')
```
