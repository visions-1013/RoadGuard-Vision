# 03 训练 YOLO11 与 YOLO26

四个模型使用相同数据划分和尽可能一致的超参数。如降低 `batch`，请在记录中保留实际值。

```python
from pathlib import Path
import json
from ultralytics import YOLO

DATA_YAML = '../data/data.yaml'
MODEL_SPECS = {'YOLO11n': 'yolo11n.pt', 'YOLO11s': 'yolo11s.pt', 'YOLO26n': 'yolo26n.pt', 'YOLO26s': 'yolo26s.pt'}
TRAIN_ARGS = {'imgsz': 640, 'epochs': 50, 'batch': 8, 'seed': 42, 'device': 0}
RUN_TRAINING = False  # 确认数据和 GPU 环境后改为 True
TRAIN_ARGS
```

```python
records = []
if RUN_TRAINING:
    for model_name, pretrained_id in MODEL_SPECS.items():
        model = YOLO(pretrained_id)
        result = model.train(data=DATA_YAML, project='../runs/train', name=model_name, **TRAIN_ARGS)
        best_weight = Path(result.save_dir) / 'weights' / 'best.pt'
        records.append({'model': model_name, 'pretrained_id': pretrained_id, 'best_weight': str(best_weight), **TRAIN_ARGS})
    Path('../runs').mkdir(exist_ok=True)
    Path('../runs/training_records.json').write_text(json.dumps(records, ensure_ascii=False, indent=2), encoding='utf-8')
records
```
